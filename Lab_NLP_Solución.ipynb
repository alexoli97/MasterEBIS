**Celda 5: Tokenizador Simple y SimpleVocabulary**

Para usar AG News con nuestras representaciones necesitamos:
1. Un **tokenizador** — divide texto en palabras.
2. Un **vocabulario** — asigna un índice entero a cada palabra.
3. Una función **index_sequence** — convierte una lista de palabras en lista de índices.

````python
def simple_tokenizer(text):
    '''
    Toma un texto y devuelve una lista de palabras individuales
    Args:
        text (str): El texto a tokenizar
    Returns ([str]): Lista de palabras
    '''
    return text.split()

class SimpleVocabulary():

    def __init__(self, corpus):
        '''
        Toma un corpus, lo tokeniza y crea diccionarios para índices y palabras
        Args:
            corpus ([str]): Lista de textos no tokenizados
        '''
        self.token2index = {}
        self.index2token = []
        current_index = 0
        for text in corpus:
            for token in simple_tokenizer(text):
                if token not in self.token2index:
                    self.token2index[token] = current_index
                    self.index2token.append(token)
                    current_index += 1

    def get_index(self, token):
        '''
        Toma una palabra y devuelve su índice
        Args:
            token (str): Una palabra
        Returns (int): Índice de la palabra
        '''
        return self.token2index.get(token, -1)  # Devuelve -1 si no está

    def get_token(self, index):
        '''
        Toma un índice y devuelve la palabra correspondiente
        Args:
            index (int): Un índice
        Returns (str): La palabra correspondiente
        '''
        if index < len(self.index2token):
            return self.index2token[index]
        return "<UNK>"

    def __len__(self):
        '''
        Devuelve el número de palabras en el vocabulario
        '''
        return len(self.index2token)

def index_sequence(sequence, vocabulary):
    '''
    Toma una lista de palabras y un vocabulario, y devuelve una lista de índices
    Args:
        sequence ([str]): Lista de palabras
        vocabulary (SimpleVocabulary): El vocabulario
    Returns ([int]): Lista de índices
    '''
    return [vocabulary.get_index(token) for token in sequence]


# Construir el vocabulario AG News a partir del corpus de entrenamiento
ag_vocab_corpus = [text for _, text in ag_train]
vocab = SimpleVocabulary(ag_vocab_corpus)

# Funciones de conversión texto <-> tensor para AG News
def tensor2text(tensor):
    return ' '.join([vocab.index2token[index] for index in tensor if 0 <= index < len(vocab.index2token)])

def text2tensor(text):
    tensor = torch.Tensor([vocab.get_index(token) for token in text.split()])
    return tensor.int()

print(f'Tamaño del vocabulario AG News: {len(vocab)} palabras')
````

## Bag of Words (BoW)

BoW representa un texto como un vector donde cada posición corresponde a una palabra del vocabulario.
El valor en cada posición es el **número de veces** que esa palabra aparece en el texto.

Propiedades clave:
- Dos textos con las mismas palabras tienen la misma representación BoW (sin importar el orden).
- El vector tiene longitud igual al vocabulario — puede ser muy grande.
- Es una representación dispersa (*sparse*): la mayoría de posiciones son 0.

**Celda 7: Implementación del BoW Embedder**

````python
def bow_embedder(tensor, vocab_length):
    '''
    Toma un tensor y el tamaño del vocabulario, y devuelve la representación BoW
    Args:
        tensor (torch.Tensor): Tensor de palabras representadas por su índice en el vocabulario
        vocab_length (int): Número de palabras en el vocabulario
    Returns (torch.Tensor): Tensor con la representación BoW
    '''
    bow = torch.zeros(vocab_length)
    for index in tensor:
        idx = int(index.item())
        if 0 <= idx < vocab_length:
            bow[idx] += 1
    return bow
````

**Celda 9: Creación del BoWDataset y DataLoaders**

**Ejercicio 1.1:** Completa la clase `BoWDataset` para cargar los textos codificados con BoW.

**Pista:** El dataset AG News (`ag_train`) es una lista de tuplas `(etiqueta, texto)`. El `embedder` toma un texto y devuelve su representación BoW.

````python
# Creamos nuestro propio dataset para cargar los textos embebidos con BoW
class BoWDataset(torch.utils.data.Dataset):
    def __init__(self, dataset, embedder):
        self.dataset = dataset
        self.embedder = embedder

    def __getitem__(self, index):
        label = self.dataset[index][0]
        text  = self.dataset[index][1]
        # Tokenizar, indexar y embeber el texto
        tokens = simple_tokenizer(text)
        indexed = torch.Tensor(index_sequence(tokens, vocab)).int()
        return label, self.embedder(indexed, len(vocab))

    def __len__(self):
        return len(self.dataset)

bow_train_data = BoWDataset(ag_train, bow_embedder)
bow_test_data = BoWDataset(ag_test, bow_embedder)
bow_trainloader = DataLoader(bow_train_data, batch_size=1000, shuffle=True)
bow_testloader = DataLoader(bow_test_data, batch_size=1000, shuffle=False)
````

**Celda 10: Red Neuronal para BoW**

**Ejercicio 1.2:** Define una red neuronal (`network`) que:
- tome un vector BoW de tamaño `len(vocab)` como entrada.
- produzca 4 logits (uno por clase) como salida.

**Pista:** Usa `nn.Sequential` con capas `nn.Linear` y `nn.ReLU`.

````python
network = nn.Sequential(
    nn.Linear(len(vocab), 256),
    nn.ReLU(),
    nn.Linear(256, 128),
    nn.ReLU(),
    nn.Linear(128, 4)
)
````

**Celda 11: Optimizador, Función de Pérdida y Entrenamiento BoW**

**Ejercicio 1.3:** Completa el ciclo de entrenamiento.

**Pista:** Pasos típicos por lote: forward → pérdida → backward → step → zero_grad.

````python
# Optimizador Adam
optimizer = optim.Adam(network.parameters())

# Función de pérdida para clasificación multiclase
loss_function = nn.CrossEntropyLoss()

epochs = 1  # El dataset es grande; una época es suficiente

for epoch in range(epochs):
    for batch_nr, (labels, data) in enumerate(bow_trainloader):
        # Poner gradientes a cero
        optimizer.zero_grad()
        # Forward pass
        prediction = network(data)
        # Calcular pérdida
        loss = loss_function(prediction, labels)
        # Backward + actualizar
        loss.backward()
        optimizer.step()
        print(
            f'\rEpoch {epoch+1} [{batch_nr+1}/{len(bow_trainloader)}] - Loss: {loss.item():.4f}',
            end=''
        )
    print()
````

**Celda 12: Evaluación del Modelo BoW (Precisión y Matriz de Confusión)**

````python
import itertools as it

def plot_confusion_matrix(cm, classes,
                          normalize=False,
                          title='Matriz de Confusión',
                          cmap=plt.cm.Blues):
    """Imprime y grafica la matriz de confusión."""
    if normalize:
        cm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]
        print("Matriz de confusión normalizada")
    else:
        print('Matriz de confusión, sin normalizar')
    print(cm)
    plt.imshow(cm, interpolation='nearest', cmap=cmap)
    plt.title(title)
    plt.colorbar()
    tick_marks = np.arange(len(classes))
    plt.xticks(tick_marks, classes, rotation=45)
    plt.yticks(tick_marks, classes)
    fmt = '.2f' if normalize else 'd'
    thresh = cm.max() / 2.
    for i, j in it.product(range(cm.shape[0]), range(cm.shape[1])):
        plt.text(j, i, format(cm[i, j], fmt),
                 horizontalalignment="center",
                 color="white" if cm[i, j] > thresh else "black")
    plt.ylabel('Etiqueta Verdadera')
    plt.xlabel('Etiqueta Predicha')
    plt.tight_layout()
````

````python
with torch.no_grad():
    correct = 0
    total = 0
    y_pred = []
    y_true = []
    for batch_nr, (labels, data) in enumerate(bow_testloader):
        outputs = network(data)
        _, predicted = torch.max(outputs, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()
        y_pred.extend(predicted.tolist())
        y_true.extend(labels.tolist())
    print(f'Precisión BoW: {100*correct/total:.2f}%')

cnf_matrix = confusion_matrix(y_true, y_pred, labels=[0, 1, 2, 3])
numpy.set_printoptions(precision=2)
plt.figure()
plot_confusion_matrix(cnf_matrix, classes=['Mundo', 'Deportes', 'Negocios', 'Ciencia/Tec'],
                      title='Matriz de confusión BoW')
plt.show()
````

## Sección 2: Word2Vec — Embeddings que Capturan Significado

BoW tiene dos limitaciones importantes:
- **Ignora el orden** de las palabras.
- **El vector crece** con el vocabulario — puede ser enorme (>100K dimensiones).
- **No captura semántica**: "gato" y "perro" son vectores ortogonales en BoW.

**Word2Vec** aprende vectores densos y pequeños (p.ej. 50–300 dimensiones) donde palabras similares tienen vectores cercanos en el espacio.

La idea central: **una palabra se define por sus vecinos**. Si "gato" y "perro" suelen aparecer en contextos similares (comen, juegan, duermen), sus vectores serán similares.

### Incrustación Skip-gram

Skip-gram entrena una red para predecir las palabras del entorno dado una palabra central.

**Celda 16: Hiperparámetros para Skip-gram**

````python
# Parámetros para el Embedding Skip-gram
WINDOW_SIZE = 2    # Número de palabras a cada lado de la palabra central
EMBEDDING_SIZE = 2 # Dimensiones del espacio de embedding (2D para visualizar)
````

**Celda 17: Función Skip-grams**

````python
import random

# Genera pares (contexto, objetivo) para entrenar Skip-gram
def skipgrams(sequences, window_size=2, shuffle=True):
    couples = []
    for sequence in sequences:
        for i, wi in enumerate(sequence):
            window_start = max(0, i - window_size)
            window_end = min(len(sequence), i + window_size + 1)
            for j in range(window_start, window_end):
                if j != i:
                    wj = sequence[j]
                    couples.append([wi, wj])
    if shuffle:
        random.shuffle(couples)
    return torch.Tensor(couples)
````

**Celda 18: Ejemplo de Generación de Skip-grams**

Veamos cómo se generan los pares de Skip-gram a partir de una oración de ejemplo.

````python
# Creamos el vocabulario
animal_vocab = SimpleVocabulary(animal_corpus)

# Oración de prueba
test_sequence = 'el cat sat en el mat'
print('La oración:', test_sequence)

# Tokenizar e indexar la oración
tokenized_sequence = simple_tokenizer(test_sequence)
indexed_sequence = [animal_vocab.get_index(token) for token in tokenized_sequence]
print('La oración indexada:', indexed_sequence)

# Generar los pares de Skip-gram sin mezclar
skipgrams_sequence = skipgrams([indexed_sequence], shuffle=False).int()
print('Todos los pares de Skip-gram:', skipgrams_sequence)

# Convertir índices a palabras
skipgrams_words = [(animal_vocab.get_token(x[0]), animal_vocab.get_token(x[1])) for x in skipgrams_sequence]
print('Los pares en texto:', skipgrams_words)

# Generar todos los pares para el corpus completo
indexed_text = [[animal_vocab.get_index(token) for token in simple_tokenizer(text)] for text in animal_corpus]
pairs = skipgrams(indexed_text, WINDOW_SIZE)
print(f'\nHay {len(pairs)} pares de Skip-gram en el corpus completo.')
````

**Ejercicio 2.1:** Crea la clase `SkipgramEmbedder`, una red neuronal que comprime la representación de una palabra a un espacio de menor dimensión y luego predice la palabra objetivo.

**Celda 19: Implementación de SkipgramEmbedder**

````python
# Una capa de incrustación utilizada para convertir un int en one-hot
to_onehot = nn.Embedding.from_pretrained(torch.eye(len(animal_vocab)))

class SkipgramEmbedder(nn.Module):
    def __init__(self):
        super().__init__()
        self.input_layer = nn.Linear(len(animal_vocab), EMBEDDING_SIZE)
        self.output_layer = nn.Linear(EMBEDDING_SIZE, len(animal_vocab))
        self.softmax = nn.Softmax(dim=-1)

    def embed(self, context_index):
        # Convertir índice a representación one-hot
        context_onehot = to_onehot(context_index.long()).float()
        # Comprimir a espacio de embedding
        embedding = self.input_layer(context_onehot)
        return embedding

    def forward(self, context_index):
        embedding = self.embed(context_index)
        output = self.output_layer(embedding)
        prediction = self.softmax(output)
        return prediction
````

#### Aprendiendo las incrustaciones

Entrena el modelo Skip-gram para que aprenda representaciones de las palabras.

**Ejercicio 2.2:** Implementa el ciclo de entrenamiento usando los pares Skip-gram.

**Pista:** Es similar al ejercicio 1.3, pero ahora usamos *pares* — el contexto como entrada y la palabra objetivo como etiqueta.

**Celda 20: Entrenamiento del Modelo SkipgramEmbedder**

````python
model = SkipgramEmbedder()
optimizer = torch.optim.RMSprop(model.parameters(), lr=0.01)
loss_function = nn.CrossEntropyLoss()

TOTAL_EPOCHS = 100
BATCH_SIZE = 64

num_batches = max(1, len(pairs) // BATCH_SIZE)

for epoch in range(TOTAL_EPOCHS):
    epoch_loss = 0
    start = 0
    for batch_nr, end in enumerate(range(BATCH_SIZE, len(pairs)+1, BATCH_SIZE)):
        end = min(end, len(pairs))
        batch_pairs = pairs[start:end]
        contexts = batch_pairs[:, 0].long()
        targets  = batch_pairs[:, 1].long()
        start = end

        # Forward
        predictions = model(contexts)
        loss = loss_function(predictions, targets)

        # Backward
        loss.backward()
        optimizer.step()
        optimizer.zero_grad()

        epoch_loss += loss.item()
        print(
            f'\rEpoch {epoch+1} [{batch_nr+1}/{num_batches}] - Loss: {loss.item():.4f}',
            end=''
        )
    print()
````

### Entrenar una FFNN con Embeddings GloVe

Usaremos los embeddings GloVe en lugar de BoW para clasificar AG News.
La diferencia clave: el vector de una noticia ya no es disperso — es un vector denso de 50 dimensiones (promedio de los embeddings de sus palabras).

**Celda 25: Implementación de w2vDataset**

````python
# Dataset que carga los textos incrustados con GloVe
class w2vDataset(torch.utils.data.Dataset):
    def __init__(self, dataset, embedder):
        self.dataset = dataset
        self.embedder = embedder

    def __getitem__(self, index):
        label = self.dataset[index][0]
        text  = self.dataset[index][1]
        tokens = simple_tokenizer(text)
        indexed = torch.Tensor(index_sequence(tokens, vocab)).int()
        return label, self.embedder(indexed, wv.vector_size)

    def __len__(self):
        return len(self.dataset)
````

**Celda 26: Creación de w2v_embedder**

El embedder codifica cada palabra con `wv[token]` y luego **promedia** los vectores para obtener una representación de longitud fija.

**Ejercicio 3.1:** Implementa `w2v_embedder`.

**Pista:** Para cada token del texto, obtén su vector GloVe. Si la palabra no está en el vocabulario preentrenado, usa un vector de ceros. Luego promedia todos los vectores.

````python
def w2v_embedder(tensor, embedding_length):
    '''
    Toma un tensor de índices y devuelve el embedding GloVe promedio del texto.
    Args:
        tensor (torch.Tensor): Índices de palabras en el vocabulario AG News
        embedding_length (int): Dimensión del embedding (50 para GloVe-50)
    Returns (torch.Tensor): Vector embedding promedio
    '''
    tensor = tensor.long()
    embedding = []
    words = [vocab.index2token[idx] for idx in tensor if 0 <= idx < len(vocab.index2token)]
    for token in words:
        if token in wv:
            embedding.append(wv[token])
        else:
            embedding.append(numpy.zeros(embedding_length))
    if len(embedding) == 0:
        return torch.zeros(embedding_length)
    embedding = numpy.array(embedding).mean(axis=0)
    return torch.tensor(embedding, dtype=torch.float32)
````

## Sección 2.1: Fundamentos de Transformers y GPT-2 (Resuelto, no hace falta cambiar nada)

- **Transformers**: Introducidos en 2017, se basan en mecanismos de atención que permiten al modelo ver toda la secuencia a la vez, sin procesar token por token como las RNNs.

- **GPT-2**: Modelo de lenguaje que predice el siguiente token dada una secuencia de entrada. Usa una pila de bloques Transformer con atención *enmascarada* (solo mira hacia atrás, no hacia adelante).

- **Auto-atención**: Permite que cada token "atienda" a todos los demás tokens anteriores, capturando dependencias a larga distancia.

**Celda 0: Preparación de los Datos**

### Preparación de los Datos

Usaremos el mismo corpus "Cat on the mat" del Ejercicio 1.

````python
animals = ['cat', 'dog', 'goat', 'elephant', 'eagle', 'zebra', 'rhino', 'hippo']
actions = ['sat', 'stood', 'jumped', 'slept']
furniture = ['mat', 'rug', 'sofa', 'bed']

# Generar el corpus
animal_corpus = [f'the {x[0]} {x[1]} in the {x[2]}' for x in itertools.product(animals, actions, furniture)]

# Funciones básicas
def simple_tokenizer(text):
    return text.split(' ')

for oracion in animal_corpus[:10]:
  print(oracion)
````

- **animal_corpus**: Todas las combinaciones posibles de animal + acción + mueble.
- **SimpleVocabulary**: Crea el vocabulario y codifica las oraciones (reutilizada del Ejercicio 1).

**Celda 1: Definición de Tokens Especiales y Extensión del Vocabulario**

````python
# CELDA 2: Definición del Vocabulario
# Definición de Tokens Especiales
PAD_TOKEN = '<pad>'
UNK_TOKEN = '<unk>'

class SimpleVocabulary():
    def __init__(self, corpus):
        if isinstance(corpus, list):
            corpus = ' '.join(corpus)

        # Obtener todas las palabras únicas del corpus
        all_words = set(word.lower() for word in simple_tokenizer(corpus))

        self.token2index = {}
        self.index2token = []

        # Agregar tokens especiales primero
        special_tokens = ['<pad>', '<unk>']
        for token in special_tokens:
            self.token2index[token] = len(self.token2index)
            self.index2token.append(token)

        # Agregar el resto del vocabulario
        for word in sorted(all_words):  # Ordenar para consistencia
            if word not in self.token2index:
                self.token2index[word] = len(self.token2index)
                self.index2token.append(word)

    def get_index(self, token):
        token = str(token).lower()
        return self.token2index.get(token, self.token2index['<unk>'])

    def get_token(self, index):
        if 0 <= index < len(self.index2token):
            return self.index2token[index]
        return '<unk>'

    def __len__(self):
        return len(self.index2token)

def index_sequence(sequence, vocabulary):
    return [vocabulary.get_index(token) for token in sequence]
````

````python
animal_vocab = SimpleVocabulary(animal_corpus)
````

**Celda 2: Definición de Hiperparámetros**

````python
batch_size = 32
block_size = 8
max_iters = 400
eval_interval = 100
learning_rate = 3e-4
device = 'cuda' if torch.cuda.is_available() else 'cpu'
eval_iters = 100
n_embd = 128
n_head = 4
n_layer = 4
dropout = 0.1
vocab_size = len(animal_vocab)
````

**Celda 3: Definición de los Componentes del Transformer**

````python
import torch
import torch.nn as nn
from torch.nn import functional as F

# Definición de la Cabeza de Atención
class Head(nn.Module):
    """Una sola cabeza de auto-atención"""
    def __init__(self, head_size):
        super().__init__()
        self.key = nn.Linear(n_embd, head_size, bias=False)
        self.query = nn.Linear(n_embd, head_size, bias=False)
        self.value = nn.Linear(n_embd, head_size, bias=False)
        self.register_buffer('tril', torch.tril(torch.ones(block_size, block_size)))

        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        B, T, C = x.shape
        k = self.key(x)   # (B, T, hs)
        q = self.query(x) # (B, T, hs)
        wei = q @ k.transpose(-2, -1) * C**-0.5  # (B, T, T)
        wei = wei.masked_fill(self.tril[:T, :T] == 0, float('-inf'))
        wei = F.softmax(wei, dim=-1)
        wei = self.dropout(wei)
        v = self.value(x)  # (B, T, hs)
        out = wei @ v      # (B, T, hs)
        return out

# Definición de la Múltiple Cabeza de Atención
class MultiHeadAttention(nn.Module):
    """Múltiples cabezas de auto-atención en paralelo"""
    def __init__(self, num_heads, head_size):
        super().__init__()
        self.heads = nn.ModuleList([Head(head_size) for _ in range(num_heads)])
        self.proj = nn.Linear(head_size * num_heads, n_embd)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        out = torch.cat([h(x) for h in self.heads], dim=-1)
        out = self.dropout(self.proj(out))
        return out

# Definición de la Red FeedForward
class FeedForward(nn.Module):
    """Una simple capa lineal seguida de una no linealidad"""
    def __init__(self, n_embd):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(n_embd, 4 * n_embd),
            nn.ReLU(),
            nn.Linear(4 * n_embd, n_embd),
            nn.Dropout(dropout),
        )

    def forward(self, x):
        return self.net(x)

# Definición del Bloque Transformer
class Block(nn.Module):
    """Bloque Transformer: comunicación seguida de computación"""
    def __init__(self, n_embd, n_head):
        super().__init__()
        head_size = n_embd // n_head
        self.sa = MultiHeadAttention(n_head, head_size)
        self.ffwd = FeedForward(n_embd)
        self.ln1 = nn.LayerNorm(n_embd)
        self.ln2 = nn.LayerNorm(n_embd)

    def forward(self, x):
        x = x + self.sa(self.ln1(x))
        x = x + self.ffwd(self.ln2(x))
        return x
````

### Explicación de los componentes del Transformer:

- **Head**: Una cabeza de atención — calcula claves, consultas y valores, y aplica una máscara para que cada token solo vea el pasado.
- **MultiHeadAttention**: Varias cabezas en paralelo, combinadas proyectando de vuelta al espacio de embeddings.
- **FeedForward**: Red neuronal simple (capa oculta + ReLU) aplicada a cada posición.
- **Block**: Combina atención y feedforward con conexiones residuales y normalización de capas.

**Celda 4: Definición del Modelo GPTSimplificado**

````python
class GPTSimplificado(nn.Module):
    def __init__(self, vocab_size):
        super().__init__()
        self.token_embedding_table = nn.Embedding(vocab_size, n_embd)
        self.position_embedding_table = nn.Embedding(block_size, n_embd)
        self.blocks = nn.Sequential(*[Block(n_embd, n_head=n_head) for _ in range(n_layer)])
        self.ln_f = nn.LayerNorm(n_embd)
        self.lm_head = nn.Linear(n_embd, vocab_size)

    def forward(self, idx, targets=None):
        B, T = idx.shape
        tok_emb = self.token_embedding_table(idx)  # (B, T, C)
        pos_emb = self.position_embedding_table(torch.arange(T, device=device))  # (T, C)
        x = tok_emb + pos_emb  # (B, T, C)
        x = self.blocks(x)  # (B, T, C)
        x = self.ln_f(x)  # (B, T, C)
        logits = self.lm_head(x)  # (B, T, vocab_size)

        if targets is None:
            loss = None
        else:
            B, T, C = logits.shape
            logits = logits.view(B*T, C)
            targets = targets.view(B*T)
            loss = F.cross_entropy(logits, targets)

        return logits, loss

    def generate(self, idx, max_new_tokens):
        for _ in range(max_new_tokens):
            idx_cond = idx[:, -block_size:]
            logits, loss = self(idx_cond)
            logits = logits[:, -1, :]  # (B, C)
            probs = F.softmax(logits, dim=-1)  # (B, C)
            idx_next = torch.multinomial(probs, num_samples=1)  # (B, 1)
            idx = torch.cat((idx, idx_next), dim=1)  # (B, T+1)
        return idx
````

### Explicación del modelo GPTSimplificado:

- **token_embedding_table** y **position_embedding_table**: Convierten índices de tokens y posiciones en vectores.
- **blocks**: Pila de bloques Transformer.
- **ln_f** y **lm_head**: Normalización final y proyección a logits sobre el vocabulario.
- **forward**: Propaga la entrada por el modelo y calcula la pérdida si se dan objetivos.
- **generate**: Genera tokens nuevos de forma autorregresiva.

**Celda 5: Función para realizar Padding en las Secuencias**

````python
encoded_corpus = [index_sequence(text.split(), animal_vocab) for text in animal_corpus]

def pad_sequences(sequences, max_length, padding_value=animal_vocab.get_index('<pad>')):
    padded_sequences = []
    for seq in sequences:
        if len(seq) < max_length:
            padded_seq = seq + [padding_value] * (max_length - len(seq))
        else:
            padded_seq = seq[:max_length]
        padded_sequences.append(padded_seq)
    return padded_sequences

# Determinar la Longitud Máxima de las Secuencias
max_length = max(block_size, max(len(seq) for seq in encoded_corpus))
print(f'Longitud máxima de las secuencias: {max_length}')

# Realizar Padding en las Secuencias
padded_encoded_corpus = pad_sequences(encoded_corpus, max_length)

# Convertir a Tensor
data = torch.tensor(padded_encoded_corpus, dtype=torch.long)

# División en Entrenamiento y Validación
n = int(0.9 * len(data))
train_data = data[:n]
val_data = data[n:]
````

**Celda 6: Función para Obtener Lotes de Datos**

````python
def get_batch(split):
    """
    Genera un lote de datos para entrenamiento o validación.

    Args:
        split (str): 'train' o 'val' para indicar el conjunto de datos.

    Returns:
        Tuple[torch.Tensor, torch.Tensor]: Tensores de entrada (x) y objetivos (y) de forma (batch_size, block_size).
    """
    data_split = train_data if split == 'train' else val_data
    batch_x = []
    batch_y = []

    for _ in range(batch_size):
        # Seleccionar una oración aleatoria
        sentence_idx = torch.randint(len(data_split), (1,)).item()
        sentence = data_split[sentence_idx]

        # Obtener la secuencia de entrada y la secuencia objetivo
        x = sentence[:block_size]
        y = sentence[1:block_size + 1]

        # Verificar la longitud de y y rellenar si es necesario
        if len(y) < block_size:
            pad_length = block_size - len(y)
            pad_tensor = torch.tensor([animal_vocab.get_index(PAD_TOKEN)] * pad_length, dtype=torch.long)
            y = torch.cat([y, pad_tensor])

        batch_x.append(x)
        batch_y.append(y)

    x = torch.stack(batch_x)
    y = torch.stack(batch_y)
    x, y = x.to(device), y.to(device)

    assert x.shape == (batch_size, block_size), f"x shape: {x.shape}"
    assert y.shape == (batch_size, block_size), f"y shape: {y.shape}"

    return x, y
````

- **get_batch**: Selecciona aleatoriamente secuencias de longitud `block_size`, desplazando los objetivos una posición (el modelo predice el siguiente token).

**Celda 7: Función para Estimar la Pérdida**

````python
@torch.no_grad()
def estimate_loss():
    """
    Calcula la pérdida promedio en los conjuntos de entrenamiento y validación.

    Returns:
        dict: Diccionario con las pérdidas promedio para 'train' y 'val'.
    """
    out = {}
    model.eval()
    for split in ['train', 'val']:
        losses = torch.zeros(eval_iters)
        for k in range(eval_iters):
            X, Y = get_batch(split)
            logits, loss = model(X, Y)
            losses[k] = loss.item()
        out[split] = losses.mean()
    model.train()
    return out
````

- **estimate_loss**: Calcula la pérdida promedio en entrenamiento y validación sin actualizar pesos (`torch.no_grad()`).

**Celda 8: Inicialización del Modelo y Parámetros**

- **Parámetros**: Tamaño de batch, longitud de bloque, iteraciones, tasa de aprendizaje, dimensiones de embeddings, cabezas de atención, capas y dropout.

**Celda 9: Ciclo de Entrenamiento del Modelo**

````python
model = GPTSimplificado(vocab_size)
model = model.to(device)
optimizer = torch.optim.AdamW(model.parameters(), lr=learning_rate)

for iter in range(max_iters):
    # Evaluar periódicamente la pérdida en entrenamiento y validación
    if iter % eval_interval == 0:
        losses = estimate_loss()
        print(f"Iteración {iter}: Pérdida Entrenamiento {losses['train']:.4f}, Pérdida Validación {losses['val']:.4f}")

    # Obtener un lote de datos de entrenamiento
    xb, yb = get_batch('train')

    # Evaluar la pérdida
    logits, loss = model(xb, yb)

    # Retropropagación
    optimizer.zero_grad(set_to_none=True)
    loss.backward()
    optimizer.step()
````

- **Entrenamiento**: En cada iteración se evalúa la pérdida periódicamente, se obtiene un lote, se calcula la pérdida, se retropropaga y se actualizan los pesos.

**Celda 10: Definición de la Función generate_text**

````python
def generate_text(model, start_sentence, max_new_tokens):
    '''
    Genera texto a partir de una oración inicial usando el modelo GPTSimplificado.

    Args:
        model (GPTSimplificado): El modelo entrenado.
        start_sentence (str): La oración inicial para la generación.
        max_new_tokens (int): Número de tokens a generar.

    Returns:
        str: El texto generado.
    '''
    # Tokenizar el contexto inicial
    context_tokens = [animal_vocab.get_index(token) for token in start_sentence.split()]

    # Convertir a tensor y añadir una dimensión de batch
    context_tensor = torch.tensor([context_tokens], dtype=torch.long).to(device)

    # Generar nuevos tokens
    generated_indices = model.generate(context_tensor, max_new_tokens=max_new_tokens)

    # Convertir los índices generados de vuelta a tokens
    generated_tokens = [animal_vocab.get_token(idx) for idx in generated_indices[0].tolist()]

    # Unir los tokens en una sola cadena de texto
    generated_text = ' '.join(generated_tokens)

    return generated_text
````
