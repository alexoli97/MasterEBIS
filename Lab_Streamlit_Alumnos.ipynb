````python
%%writefile dashboard.py
import streamlit as st
import os
import seaborn as sns
import pandas as pd
import plotly.express as px

# set_page_config DEBE ser la primera llamada a Streamlit
st.set_page_config(
    page_title="Dashboard de Pingüinos",
    page_icon="🐧",
    layout="wide"
)

# @st.cache_data evita recargar los datos en cada interacción
@st.cache_data
def cargar_datos():
    df = sns.load_dataset('penguins')
    df = df.dropna()
    return df

df = cargar_datos()
````

## Sección 5: Visualizaciones con Plotly *(código dado — ejecútalo y explóralo)*

Las visualizaciones son complejas, así que el código está completo. Tu tarea es **ejecutar cada celda y entender qué hace cada parámetro**.

**Celda 10: Histograma y Boxplot**

````python
%%writefile -a dashboard.py

st.divider()
st.subheader("📈 Distribuciones")

nombres_variables = {
    'bill_length_mm':    'Longitud del pico (mm)',
    'bill_depth_mm':     'Profundidad del pico (mm)',
    'flipper_length_mm': 'Longitud de la aleta (mm)',
    'body_mass_g':       'Masa corporal (g)'
}

col_izq, col_der = st.columns(2)

with col_izq:
    variable_hist = st.selectbox(
        label="Variable para el histograma",
        options=list(nombres_variables.keys()),
        format_func=lambda k: nombres_variables[k]
    )
    fig_hist = px.histogram(
        df_filtrado, x=variable_hist, color='species',
        nbins=30, barmode='overlay', opacity=0.7,
        title="Distribución por especie",
        labels={variable_hist: nombres_variables[variable_hist], 'species': 'Especie'}
    )
    st.plotly_chart(fig_hist, use_container_width=True)

with col_der:
    variable_box = st.selectbox(
        label="Variable para el diagrama de caja",
        options=list(nombres_variables.keys()),
        index=2,
        format_func=lambda k: nombres_variables[k],
        key="boxplot_var"   # key único para evitar conflicto con el selectbox anterior
    )
    fig_box = px.box(
        df_filtrado, x='species', y=variable_box,
        color='species', points='outliers',
        title="Distribución por especie (Boxplot)",
        labels={'species': 'Especie', variable_box: nombres_variables[variable_box]}
    )
    fig_box.update_layout(showlegend=False)
    st.plotly_chart(fig_box, use_container_width=True)
````

**Celda 11: Scatter Plot y Composición por Isla**

````python
%%writefile -a dashboard.py

st.divider()
st.subheader("🔵 Relaciones entre Variables")

col_sx, col_sy = st.columns(2)
with col_sx:
    eje_x = st.selectbox("Eje X", list(nombres_variables.keys()),
                          format_func=lambda k: nombres_variables[k], key="scatter_x")
with col_sy:
    eje_y = st.selectbox("Eje Y", list(nombres_variables.keys()), index=2,
                          format_func=lambda k: nombres_variables[k], key="scatter_y")

fig_scatter = px.scatter(
    df_filtrado, x=eje_x, y=eje_y,
    color='species', symbol='sex', size='body_mass_g',
    hover_data=['island'],
    title=f"{nombres_variables[eje_x]} vs {nombres_variables[eje_y]}",
    labels={eje_x: nombres_variables[eje_x], eje_y: nombres_variables[eje_y]}
)
st.plotly_chart(fig_scatter, use_container_width=True)

st.divider()
st.subheader("🏝️ Composición por Isla")

col_bar, col_pie = st.columns(2)
with col_bar:
    conteo_isla = df_filtrado.groupby(['island', 'species']).size().reset_index(name='conteo')
    fig_bar = px.bar(conteo_isla, x='island', y='conteo', color='species',
                     barmode='stack', title="Pingüinos por isla y especie",
                     labels={'island': 'Isla', 'conteo': 'Número', 'species': 'Especie'})
    st.plotly_chart(fig_bar, use_container_width=True)

with col_pie:
    conteo_especie = df_filtrado['species'].value_counts().reset_index()
    conteo_especie.columns = ['Especie', 'Conteo']
    fig_pie = px.pie(conteo_especie, names='Especie', values='Conteo',
                     title="Proporción de especies", hole=0.4)
    st.plotly_chart(fig_pie, use_container_width=True)
````

**Celda 12: Mapa de Correlaciones y Estadísticas**

````python
%%writefile -a dashboard.py

st.divider()
st.subheader("🌡️ Mapa de Correlaciones")

columnas_numericas = ['bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'body_mass_g']
etiquetas_cortas = {
    'bill_length_mm': 'Long. pico', 'bill_depth_mm': 'Prof. pico',
    'flipper_length_mm': 'Long. aleta', 'body_mass_g': 'Masa corporal'
}
matriz_corr = df_filtrado[columnas_numericas].corr()
matriz_corr.index = [etiquetas_cortas[c] for c in columnas_numericas]
matriz_corr.columns = [etiquetas_cortas[c] for c in columnas_numericas]

fig_heatmap = px.imshow(matriz_corr, text_auto='.2f',
                         color_continuous_scale='RdBu_r', zmin=-1, zmax=1,
                         title="Correlación de Pearson", aspect='auto')
st.plotly_chart(fig_heatmap, use_container_width=True)

st.subheader("📋 Estadísticas Descriptivas")
stats = df_filtrado[columnas_numericas].describe().round(2)
stats.index = ['Conteo', 'Media', 'Desv. Est.', 'Mínimo', 'Q1', 'Mediana', 'Q3', 'Máximo']
stats.columns = [etiquetas_cortas[c] for c in columnas_numericas]
st.dataframe(stats, use_container_width=True)

st.divider()
st.caption("Datos: Palmer Penguins Dataset (Gorman et al., 2014) | Máster en IA / Deep Learning")
````
